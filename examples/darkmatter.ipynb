### GWLENS

Dark Matter is a fundamental ingrediant in Standard Model of Cosmology. They are also the backbone of large scale structures. In Cold Dark Matter paradigm, a commonly used model for Halos is the Navarro Frenk White density profile. The profile has two parameters $\rho_s$ and $r_s$ known as the characteristic density and scale radius respectively.


GWLENS is packed with tools to generate MFW lens amplification factor in the WO limit and also the other parameters connected to NFW profile like the computation of $r_s$, $\rho_s$, $r_{200}$, $M_{200}$  and other relevant quantities needed to study halos.
