### **GWLENS**

Dark Matter is a fundamental ingrediant in Standard Model of Cosmology. They are also the backbone of large scale structures. In Cold Dark Matter paradigm, a commonly used model for Halos is the Navarro Frenk White density profile. The profile has two parameters $\rho_s$ and $r_s$ known as the characteristic density and scale radius respectively.


GWLENS is packed with tools to generate NFW lens amplification factor in the WO limit and also the other parameters connected to NFW profile like the computation of $r_s$, $\rho_s$, $r_{200}$, $M_{200}$  and other relevant quantities needed to study halos. It does also support *dressed primordial black holes* (dPBH) and bare PBHs (through point mass lens). The dPBH implementation is from the paper (https://arxiv.org/pdf/2303.17601) 

### NFW Halo tools

**GWLENS** features a halo class which helps calculate the halo params like virial Mass $M_{vir}$, concentration- $c_v$, $r_{200}$ .

```python
from gwlens.tools import halo

M200 = 1e+6

#help(halo) #uncomment to see the halo class documentation

#Initialise the halo class with M200 and redshift of the lens
darkhalo = halo.NFW(M200,zl=1.0)

darkhalo.summary()
```
