### Guideline to implement new lens model into the existing structure


 - Create a python script for the desired new lens model in the `lens` folder. The name of the script should follow the name of the lens_model (eq. `gwlens/lens/<lens_model_name>.py`)
 



 #### Checklists

  - Make sure the you lens model name is included in the valid models list in `interpolator.py`
