```python
from gwlens import solver1d
import matplotlib.pyplot as plt
import numpy as np
#from gwlens.tools import plotting


rs = 2.0
y = 6.0

lensmodel = solver1d.nfw(k=rs)
lens = solver1d.Potential(lensmodel)

w = np.logspace(-3,2,1000)


F = solver1d.amplification_many(w,y,lens)

plt.semilogx(w,np.abs(F))
plt.xlabel("w")
plt.ylabel("$|F(w)|$")
```
