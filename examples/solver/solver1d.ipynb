```python
from gwlens import solver1d
import matplotlib.pyplot as plt
import numpy as np
import time
#from gwlens.tools import plotting


rs = 3.0
y = 1.0

lensmodel = solver1d.nfw(k=rs)
lens = solver1d.Potential(lensmodel)

w = np.logspace(-3,3,20000)
t = time.perf_counter()

F = solver1d.amplification_many(w,y,lens,interpolate=False,n_workers=2)
print(f"many w : {w.size} frequencies at one y          {time.perf_counter() - t:.2f} s")

plt.semilogx(w,np.abs(F))
plt.xlabel("w")
plt.ylabel("$|F(w)|$")
```

```python
def my_psi(x):
    return 0.5 * np.log(x**2 + 1.0)

lens1 = solver1d.Potential(my_psi)
F1 = solver1d.amplification_many(w,1.0,lens1)

plt.semilogx(w,np.abs(F1))
```
