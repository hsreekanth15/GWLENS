```python
from gwlens.lens import cis
import numpy as np
import matplotlib.pyplot as plt
from mpmath import mp
from gwlens.tools import plotting

#plotting.apply_figure_style()

```

```python
mp.dps = 10  # set precision to 50 decimal places (arbitrary precision)

w = np.logspace(-3, 2, 2000)
y = 0.1
core = 0.1
result = cis.CISparallel(w, y, core, processes=4)
```

```python
#plt.semilogx(w, result.real, label='Real part')
#plt.semilogx(w, result.imag, label='Imaginary part')
plt.semilogx(w, np.abs(result), label='Magnitude')
plt.xlabel('w')
plt.ylabel('CIS Amplification')
plt.title('CIS Amplification vs w for y=0.1 and core=0.1')
plt.legend()
plt.grid()
plt.show()
```
